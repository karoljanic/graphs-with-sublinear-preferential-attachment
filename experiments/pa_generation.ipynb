```python
from src.preferential_attachment_graph import PreferentialAttachmentGraph, FitnessType
from src.drawer import Drawer
```

```python
G = PreferentialAttachmentGraph(n=200, m=3, fitness=FitnessType.POLY, fitness_alpha=1, track_history=False, select_planar_subgraph=True)

print(f"Number of nodes: {G.number_of_nodes()}"
      f"\nNumber of edges: {G.number_of_edges()}"
      f"\nAverage degree: {sum(dict(G.degree()).values()) / G.number_of_nodes():.2f}")

drawer = Drawer()
drawer.draw_graph(G, title="Preferential Attachment Graph with Sublinear Fitness")
```

```python
# draw degree distribution
import matplotlib.pyplot as plt

degree_sequence = sorted([d for n, d in G.degree()], reverse=True)
degree_count = {d: degree_sequence.count(d) for d in set(degree_sequence)}
plt.bar(degree_count.keys(), degree_count.values())
plt.title("Degree Distribution")
plt.xlabel("Degree")
plt.ylabel("Count")
plt.show()
```
